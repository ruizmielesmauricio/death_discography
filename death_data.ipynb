```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials

# Spotify authentication
auth_manager = SpotifyClientCredentials(
    client_id=SPOTIFY_CLIENT_ID,
    client_secret=SPOTIFY_CLIENT_SECRET
)

sp = spotipy.Spotify(auth_manager=auth_manager)

# Official Spotify artist ID for Death
DEATH_ARTIST_ID = "4f5V3PQ66nIrBCqugJtaGn"

# Test artist retrieval
artist_info = sp.artist(DEATH_ARTIST_ID)

print("Artist Name:", artist_info['name'])
artist_info = sp.artist(DEATH_ARTIST_ID)

print(artist_info)
```

```python
import pandas as pd
all_albums = []

offset = 0
limit = 10

while True:
    response = sp.artist_albums(
        DEATH_ARTIST_ID,
        album_type='album',
        limit=limit,
        offset=offset
    )
    
    all_albums.extend(response['items'])
    
    if response['next'] is None:
        break
    
    offset += limit

for album in all_albums:
    print(album['name'], "|", album['release_date'])
```

```python
import pandas as pd

albums_df = pd.DataFrame([
    {
        "album_name": album["name"],
        "release_date": album["release_date"],
        "total_tracks": album["total_tracks"],
        "spotify_album_id": album["id"],
        "spotify_url": album["external_urls"]["spotify"],
        "artwork_url": album["images"][0]["url"] if album["images"] else None
    }
    for album in all_albums
])

albums_df
```

```python
# Studio albums

OFFICIAL_ALBUM_KEYWORDS = [
    "Scream Bloody Gore",
    "Leprosy",
    "Spiritual Healing",
    "Human",
    "Individual Thought Patterns",
    "Symbolic",
    "The Sound of Perseverance"
]

filtered_albums = []

for official_name in OFFICIAL_ALBUM_KEYWORDS:
    matches = albums_df[
        albums_df["album_name"].str.contains(official_name, case=False, regex=False)
    ].copy()
    
    if len(matches) > 0:
        # keep the earliest release date match
        best_match = matches.sort_values("release_date").iloc[0]
        filtered_albums.append(best_match)

clean_albums_df = pd.DataFrame(filtered_albums).reset_index(drop=True)

clean_albums_df
```

```python
# Orig Tracklist
ORIGINAL_TRACKLISTS = {
    "Scream Bloody Gore": [
        "Infernal Death", "Zombie Ritual", "Denial of Life", "Sacrificial",
        "Mutilation", "Regurgitated Guts", "Baptized in Blood",
        "Torn to Pieces", "Evil Dead", "Scream Bloody Gore"
    ],
    "Leprosy": [
        "Leprosy", "Born Dead", "Forgotten Past", "Left to Die",
        "Pull the Plug", "Open Casket", "Primitive Ways", "Choke on It"
    ],
    "Spiritual Healing": [
        "Living Monstrosity", "Altering the Future", "Defensive Personalities",
        "Within the Mind", "Spiritual Healing", "Low Life", "Genetic Reconstruction",
        "Killing Spree"
    ],
    "Human": [
        "Flattening of Emotions", "Suicide Machine", "Together as One",
        "Secret Face", "Lack of Comprehension", "See Through Dreams",
        "Cosmic Sea", "Vacant Planets"
    ],
    "Individual Thought Patterns": [
        "Overactive Imagination", "In Human Form", "Jealousy", "Trapped in a Corner",
        "Nothing Is Everything", "Mentally Blind", "Individual Thought Patterns",
        "Destiny", "Out of Touch", "The Philosopher"
    ],
    "Symbolic": [
        "Symbolic", "Zero Tolerance", "Empty Words", "Sacred Serenity",
        "1,000 Eyes", "Without Judgement", "Crystal Mountain",
        "Misanthrope", "Perennial Quest"
    ],
    "The Sound of Perseverance": [
        "Scavenger of Human Sorrow", "Bite the Pain", "Spirit Crusher",
        "Story to Tell", "Flesh and the Power It Holds", "Voice of the Soul",
        "To Forgive Is to Suffer", "A Moment of Clarity", "Painkiller"
    ]
}
```

```python
tracks_data = []

for _, row in clean_albums_df.iterrows():
    
    album_id = row["spotify_album_id"]
    album_name = row["album_name"]
    release_date = row["release_date"]
    artwork_url = row["artwork_url"]
    
    tracks = sp.album_tracks(album_id)
    
    for track in tracks["items"]:
        
        tracks_data.append({
            "album_name": album_name,
            "release_date": release_date,
            "artwork_url": artwork_url,
            "track_name": track["name"],
            "track_number": track["track_number"],
            "duration_ms": track["duration_ms"],
            "spotify_track_id": track["id"]
        })

tracks_df = pd.DataFrame(tracks_data)

tracks_df.head()
```

```python
# Only original tracks
clean_tracks = []

for album_name, official_tracks in ORIGINAL_TRACKLISTS.items():
    album_tracks = tracks_df[
        tracks_df["album_name"].str.contains(album_name, case=False, regex=False)
    ].copy()
    
    for track_position, official_track in enumerate(official_tracks, start=1):
        match = album_tracks[
            album_tracks["track_name"].str.lower().str.strip() == official_track.lower().strip()
        ]
        
        if len(match) > 0:
            selected = match.iloc[0].copy()
            selected["official_track_number"] = track_position
            selected["official_track_name"] = official_track
            clean_tracks.append(selected)
        else:
            print(f"Missing track: {album_name} - {official_track}")

clean_tracks_df = pd.DataFrame(clean_tracks)

clean_tracks_df = clean_tracks_df.sort_values(
    ["release_date", "official_track_number"]
).reset_index(drop=True)

clean_tracks_df
```

```python
### Listeners data
import requests
import time

def get_lastfm_track_info(artist, track):
    url = "http://ws.audioscrobbler.com/2.0/"
    
    params = {
        "method": "track.getInfo",
        "api_key": LASTFM_API_KEY,
        "artist": artist,
        "track": track,
        "format": "json"
    }
    
    response = requests.get(url, params=params)
    data = response.json()
    
    if "track" in data:
        return {
            "lastfm_track_listeners": data["track"].get("listeners"),
            "lastfm_track_playcount": data["track"].get("playcount")
        }
    
    return {
        "lastfm_track_listeners": None,
        "lastfm_track_playcount": None
    }
```

```python
lastfm_track_data = []

for _, row in clean_tracks_df.iterrows():
    info = get_lastfm_track_info("Death", row["official_track_name"])
    
    lastfm_track_data.append(info)
    
    time.sleep(0.25)

lastfm_track_df = pd.DataFrame(lastfm_track_data)

clean_tracks_df = pd.concat(
    [clean_tracks_df.reset_index(drop=True), lastfm_track_df],
    axis=1
)

clean_tracks_df.head()
clean_tracks_df.to_csv("death_tracks_with_lastfm.csv", index=False)
```

```python
import musicbrainzngs
import requests
import time
import pandas as pd

musicbrainzngs.set_useragent(
    "death-dashboard-project",
    "1.0",
    "macob39@example.com"
)
```

```python
def get_musicbrainz_recording_id(artist, track):
    try:
        result = musicbrainzngs.search_recordings(
            artist=artist,
            recording=track,
            limit=5
        )
        
        recordings = result.get("recording-list", [])
        
        if len(recordings) > 0:
            return recordings[0]["id"]
        
        return None
    
    except Exception as e:
        print(f"MusicBrainz error for {track}: {e}")
        return None
```

```python
def get_acousticbrainz_bpm(mbid):
    if pd.isna(mbid):
        return None
    
    url = f"https://acousticbrainz.org/api/v1/{mbid}/low-level"
    
    try:
        response = requests.get(url, timeout=10)
        
        if response.status_code != 200:
            return None
        
        data = response.json()
        
        bpm = data.get("rhythm", {}).get("bpm")
        return bpm
    
    except Exception as e:
        print(f"AcousticBrainz error for {mbid}: {e}")
        return None
```

```python
musicbrainz_ids = []
bpm_values = []

for _, row in clean_tracks_df.iterrows():
    track_name = row["official_track_name"]
    
    mbid = get_musicbrainz_recording_id("Death", track_name)
    musicbrainz_ids.append(mbid)
    
    bpm = get_acousticbrainz_bpm(mbid)
    bpm_values.append(bpm)
    
    time.sleep(1.2)

clean_tracks_df["musicbrainz_recording_id"] = musicbrainz_ids
clean_tracks_df["bpm"] = bpm_values
```
